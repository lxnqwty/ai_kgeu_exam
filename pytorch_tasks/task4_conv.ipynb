```python
import torch

input_images = torch.tensor([
    [[[1, 2, 3], [1, 2, 3], [1, 2, 3]],
    [[1, 2, 3], [1, 2, 3], [1, 2, 3]],
    [[1, 2, 3], [1, 2, 3], [1, 2, 3]]]
])

# Исходная размерность изображения
input_images.shape
```

## Padding

```python
import torch.nn.functional as F


# Рytorch 
def padding2d_builtin(input_images):
    return  F.pad(input_images, (1, 1, 1, 1))


print(padding2d_builtin(input_images).shape)


# Ручная реализация 
def padding2d(input_images):
    b, c, h, w = input_images.shape

    padding_images = torch.zeros(b, c, h + 2, w + 2)
    padding_images[:, :, 1:-1, 1:-1] = input_images

    return padding_images

print(padding2d(input_images).shape)
```

## Выходная размерность

```python
# Рytorch 
def get_out_shape_builtin(input_shape, out_channels, kernel_size, stride, padding):
    _, c, _, _ = input_shape

    conv = torch.nn.Conv2d(c, out_channels, kernel_size, stride=stride, padding=padding, bias=False)
    dummy = torch.zeros(input_shape)

    with torch.no_grad():
         out = conv(dummy)

    return tuple(out.shape)

print(get_out_shape_builtin(input_images.shape, 1, 1, 2, 0))


# Ручная реализация 
def get_out_shape(input_shape, out_channels, kernel_size, stride, padding):
    b, _, h, w = input_shape

    H_out = (h - 2 * padding - kernel_size) // stride + 1
    W_out = (w - 2 * padding - kernel_size) // stride + 1

    return b, out_channels, H_out, W_out

print(get_out_shape(input_images.shape, 1, 1, 2, 0))
```
